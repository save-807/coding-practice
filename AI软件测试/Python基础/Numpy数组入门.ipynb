```python
#导包
import numpy as np
```

# 1.Numpy的属性介绍

## 1.1创建ndarray对象

```python
#numpy → 和数学操作相关的库，numpy的数组是多维的，并且是固定类型的，numpy的数组是ndarray，主要用于矩阵运行
#1.创建numpy的ndarray对象
arr1 = np.arange(15).reshape(3,5)
print(arr1)
# arr2 = np.zeros((2,3,4))
# print(arr2.ndim)
# print(arr2)
```

## 1.2.演示 numpy 的常用属性

```python
print(f'numpy的轴： {arr1.ndim}')
print(f'numpy的维度：{arr1.shape}')
print(f'numpy的元素类型： {arr1.dtype}')
print(f'numpy的元素个数：{arr1.size}')
print(f'numpy的元素占用的字节数：{arr1.itemsize}')
print(f'numpy的元素类型：{type(arr1)}')
```

# 2.创建Numpy数组的ndarray对象

## 2.1array()函数，把python列表->ndarray对象

```python
# 1.创建python列表对象
my_list=[11,22,33,44]
print(type(my_list))
# 2.把python列表->ndarray对象
arr2=np.array(my_list)
print(arr2)  # [11 22 33 44]
print(type(arr2))  # <class 'numpy.ndarray'>
```

## 2.2 arange()函数，创建ndarray对象

```python
# arange(起始，结束，步长，dtype=类型)，类似于：python中的arange()函数。
# 1.创建ndarray对象
arr3 = np.arange(0,10,2,dtype=np.int64)
# 2.打印ndarray对象
print(arr3)
print(type(arr3))
print(f'numpy的dtype类型：{arr3.dtype}')
```

## 2.3随机矩阵生成 ndarray 对象

```python
# 1.random.rand() 生成0.0~1.0之间，包括左但不包括右的 随机小数矩阵
arr4 = np.random.rand(3,5)
print(arr4)
print(type(arr4))
```

```python
# 2.random.randint(起始值，结束值，size=（行，列）)生成指定范围，包左不包右的随机证书矩阵
arr5 = np.random.randint(3,9,size=(2,6))
print(arr5)
print(type(arr5))
```

```python
# 3.random.uniform(起始值, 结束值， size=(行，列))生成指定范围，包左不包右，随机小数矩阵
arr6 = np.random.uniform(3,5,size=(4,3))
print(arr6)
print(type(arr6))
```

## 2.4 numpy的类型转换 astype()

```python
# 1.创建ndarray对象
arr7 = np.arange(0, 10, 2, dtype=np.int64)
print(arr7)
print(arr7.dtype)
```

```python
# 2. 把上述的arr7 ndarray对象的元素 从int64 -> float32
arr8 = arr7.astype(np.float32)
print(arr8)
print(arr8.dtype)
```
